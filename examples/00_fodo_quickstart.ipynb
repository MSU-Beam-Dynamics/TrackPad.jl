# 00 - FODO quick start

TrackPad uses immutable elements, an ordered `Lattice`, and a lightweight `Beam`.
Coordinates are `(x, px, y, py, z, delta_E)`; beam energy is **kinetic energy in eV**.

```julia
import Pkg
EXAMPLES_DIR = isfile(joinpath(pwd(), "common.jl")) ? pwd() : joinpath(pwd(), "examples")
Pkg.activate(EXAMPLES_DIR)
using TrackPad, StaticArrays
include(joinpath(EXAMPLES_DIR, "common.jl"))
using .TrackPadExamples

using CairoMakie
```

```julia
beam = default_beam()
ring = fodo_ring()

(element_count=length(ring), circumference=total_length(ring), periodic=isperiodic(ring))
```

```julia
r0 = @SVector [1e-3, 0.0, 0.5e-3, 0.0, 0.0, 0.0]
r1 = ringpass(ring, r0, beam, 20)
tw = periodic_twiss(ring, beam; sample_integrator_steps=true, max_step=0.1)

(tunes=(tw.tunex, tw.tuney), final_coordinates=r1)
```

```julia
fig = Figure(size=(700, 330))
ax = Axis(fig[1, 1], xlabel="s [m]", ylabel="beta [m]")
lines!(ax, tw.s, tw.betax, label="beta_x")
lines!(ax, tw.s, tw.betay, label="beta_y")
axislegend(ax)
fig
```
