# 02 - TPSA transfer map

PolySeries supplies the TPSA scalar type; TrackPad supplies element maps. `tpsa_map` returns six polynomials in canonical coordinate order.

```julia
import Pkg
EXAMPLES_DIR = isfile(joinpath(pwd(), "common.jl")) ? pwd() : joinpath(pwd(), "examples")
Pkg.activate(joinpath(EXAMPLES_DIR, "environments", "tpsa"))
Pkg.instantiate()
using TrackPad, PolySeries

include(joinpath(EXAMPLES_DIR, "common.jl")); using .TrackPadExamples
```

```julia
ring, beam = madx_fodo()
map = tpsa_map(ring, beam; order=2)
```

```julia
unit_power(j) = [i == j ? 1 : 0 for i in 1:6]
M = [element(map[i], unit_power(j)) for i in 1:6, j in 1:6]
T166 = element(map[1], [0, 0, 0, 0, 0, 2])
T116 = element(map[1], [1, 0, 0, 0, 0, 1])

(linear_map=M, selected_second_order_terms=(T166=T166, T116=T116))
```

`element(polynomial, exponents)` returns the Taylor coefficient, including the usual factorial normalization. Increase `order` only when needed: coefficient count and tracking cost grow combinatorially.
